```python
import numpy as np
import matplotlib.pyplot as plt

# 目標函數：y = x^2
def func(x):
    return np.square(x)

# 目標函數一階導數：dy/dx = 2*x
def dfunc(x):
    return 2 * x

# 梯度下降法
def GD(x_start, df, epochs, lr):
    """
    梯度下降法
    x_start：起始點
    df：目標函數的一階導數
    epochs：反覆運算次數
    lr：學習率
    """
    xs = np.zeros(epochs + 1)
    x = x_start
    xs[0] = x

    for i in range(epochs):
        dx = df(x)
        v = -dx * lr
        x += v
        xs[i + 1] = x

    return xs


# Main
# 起始權重
x_start = 5

epochs = 15

# 學習率
lr = 0.3

# 執行梯度下降法
x = GD(x_start, dfunc, epochs, lr=lr)

# 輸出每次迭代的結果
print("梯度下降法執行結果：")
print(x)

print("\n最後的 x 值：", x[-1])
print("最後的 y 值：", func(x[-1]))

# 繪製圖形
color = 'r'

from numpy import arange

t = arange(-6.0, 6.0, 0.01)

plt.figure(figsize=(8, 5))

# 藍色目標函數曲線
plt.plot(t, func(t), c='b', label='y=x^2')

# 紅色梯度下降路徑
plt.plot(x, func(x), c=color, label='lr={}'.format(lr))

# 紅色迭代位置
plt.scatter(x, func(x), c=color)

# 圖表標題及座標軸
plt.title('Gradient Descent')
plt.xlabel('x')
plt.ylabel('y')

plt.legend()
plt.grid(True)

plt.show()
```
